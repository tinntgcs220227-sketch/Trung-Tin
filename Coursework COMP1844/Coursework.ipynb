```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

plt.figure(figsize=(10, 6))

plt.title("Task1: London Underground Network Section", fontsize=14, fontweight='bold')

G = nx.Graph()
pos = {'Oxford Circus': (0, 10), 'Piccadilly Circus': (2, 5), 'Charing Cross': (4, 0),
       'Leicester Square': (4, 5), 'Covent Garden': (6, 7)}

G.add_edges_from([
    ('Oxford Circus', 'Piccadilly Circus', {'dist': 0.5, 'color': 'brown'}),
    ('Piccadilly Circus', 'Charing Cross', {'dist': 0.6, 'color': 'brown'}),
    ('Charing Cross', 'Leicester Square', {'dist': 0.4, 'color': 'black'}),
    ('Piccadilly Circus', 'Leicester Square', {'dist': 0.4, 'color': 'blue'}),
    ('Leicester Square', 'Covent Garden', {'dist': 0.2, 'color': 'blue'})
])

colors = [data['color'] for _, _, data in G.edges(data=True)]

nx.draw_networkx_nodes(G, pos, node_color='lightgrey', node_size=500)
nx.draw_networkx_edges(G, pos, edge_color=colors, width=3)

label_pos = {
    'Oxford Circus': (pos['Oxford Circus'][0], pos['Oxford Circus'][1] + 0.8),
    'Piccadilly Circus': (pos['Piccadilly Circus'][0] - 1.1, pos['Piccadilly Circus'][1]), 
    'Charing Cross': (pos['Charing Cross'][0], pos['Charing Cross'][1] - 0.9), 
    'Leicester Square': (pos['Leicester Square'][0] + 0.9, pos['Leicester Square'][1] - 0.7),
    'Covent Garden': (pos['Covent Garden'][0] + 0.5, pos['Covent Garden'][1] - 0.6)
}

nx.draw_networkx_labels(G, label_pos, font_size=10, font_weight='bold')

edge_distances = nx.get_edge_attributes(G, 'dist')

edge_labels_with_units = {edge: f"{dist} km" for edge, dist in edge_distances.items()}

nx.draw_networkx_edge_labels(
    G, pos, edge_labels=edge_labels_with_units, 
    font_size=9, font_weight='bold', 
    bbox=dict(facecolor='white', edgecolor='none', alpha=1.0, pad=2)
)

legend_elements = [
    Line2D([0], [0], color='blue', lw=3, label='Piccadilly'),
    Line2D([0], [0], color='black', lw=3, label='Northern'),
    Line2D([0], [0], color='brown', lw=3, label='Bakerloo')
]
plt.legend(handles=legend_elements, title="Key", loc='lower right', frameon=True, fontsize=10, title_fontsize=11)

plt.margins(0.2) 
plt.axis('off')
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

plt.figure(figsize=(16, 12))
plt.title("Task 2: London Underground", fontsize=18, fontweight='bold', pad=20)

G = nx.Graph()

pos = {
    
    'Oxford Circus': (5, 15), 
    'Piccadilly Circus': (10, 10), 
    'Charing Cross': (15, 5),
    'Leicester Square': (15, 10), 
    'Covent Garden': (18, 13),

   
    'Baker Street': (-5, 25), "Regent's Park": (0, 20), 'Elephant & Castle': (20, 0),

    
    'Hyde Park Corner': (0, 10), 'Green Park': (5, 10), 'Holborn': (20, 15),

    
    'Warren Street': (15, 25), 'Goodge Street': (15, 20), 
    'Tottenham Court Road': (15, 15), 'Embankment': (15, 0),

    
    'Marble Arch': (-5, 15), 'Bond Street': (0, 15), 
    'Chancery Lane': (25, 15),

    
    'Euston': (10, 30), 'Victoria': (0, 5), 'Pimlico': (0, 0)
}


G.add_edges_from([
    
    ('Baker Street', "Regent's Park", {'dist': 0.4, 'color': 'brown'}),
    ("Regent's Park", 'Oxford Circus', {'dist': 0.8, 'color': 'brown'}),
    ('Oxford Circus', 'Piccadilly Circus', {'dist': 0.5, 'color': 'brown'}), 
    ('Piccadilly Circus', 'Charing Cross', {'dist': 0.6, 'color': 'brown'}), 
    ('Charing Cross', 'Elephant & Castle', {'dist': 1.6, 'color': 'brown'}),

    
    ('Hyde Park Corner', 'Green Park', {'dist': 0.5,'color': 'blue'}),
    ('Green Park', 'Piccadilly Circus', {'dist': 0.4, 'color': 'blue'}),
    ('Piccadilly Circus', 'Leicester Square', {'dist': 0.4, 'color': 'blue'}),
    ('Leicester Square', 'Covent Garden', {'dist': 0.2, 'color': 'blue'}),     
    ('Covent Garden', 'Holborn', {'dist': 0.3, 'color': 'blue'}),

    
    ('Warren Street', 'Goodge Street', {'dist': 0.3, 'color': 'black'}),
    ('Goodge Street', 'Tottenham Court Road', {'dist': 0.4, 'color': 'black'}),
    ('Tottenham Court Road', 'Leicester Square', {'dist': 0.3, 'color': 'black'}),
    ('Leicester Square', 'Charing Cross', {'dist': 0.4, 'color': 'black'}),   
    ('Charing Cross', 'Embankment', {'dist': 0.2, 'color': 'black'}),

    
    ('Marble Arch', 'Bond Street', {'dist': 0.4, 'color': 'red'}),
    ('Bond Street', 'Oxford Circus', {'dist': 0.2, 'color': 'red'}),
    ('Oxford Circus', 'Tottenham Court Road', {'dist': 0.3, 'color': 'red'}),
    ('Tottenham Court Road', 'Holborn', {'dist': 0.4, 'color': 'red'}),
    ('Holborn', 'Chancery Lane', {'dist': 0.4, 'color': 'red'}),

    
    ('Euston', 'Warren Street', {'dist': 0.4, 'color': 'lightblue'}),
    ('Warren Street', 'Oxford Circus', {'dist': 0.9, 'color': 'lightblue'}),
    ('Oxford Circus', 'Green Park', {'dist': 0.7, 'color': 'lightblue'}),
    ('Green Park', 'Victoria', {'dist': 0.9, 'color': 'lightblue'}),
    ('Victoria', 'Pimlico', {'dist': 0.7, 'color': 'lightblue'})
])



colors = [data['color'] for _, _, data in G.edges(data=True)]


nx.draw_networkx_edges(G, pos, edge_color=colors, width=5)


transfer_stations = [node for node in G.nodes() if G.degree[node] > 2]
regular_stations = [node for node in G.nodes() if G.degree[node] <= 2]


nx.draw_networkx_nodes(
    G, pos, 
    nodelist=transfer_stations,
    node_color='white', 
    edgecolors='black', 
    linewidths=2.5, 
    node_size=350
)


nx.draw_networkx_nodes(
    G, pos, 
    nodelist=regular_stations,
    node_color='lightgrey', 
    edgecolors='none',
    node_size=250
)


label_config = {
   
    'Oxford Circus': (0, 35, 'center', 'bottom'),         
    'Piccadilly Circus': (-5, 25, 'left', 'top'),       
    'Charing Cross': (-12, -12, 'right', 'top'),          
    'Leicester Square': (12, -12, 'left', 'top'),         
    'Tottenham Court Road': (-80, 12, 'center', 'bottom'),   
    'Green Park': (12, -30, 'center', 'bottom'),           
    'Warren Street': (12, 12, 'left', 'bottom'),           

  
    'Marble Arch': (0, 15, 'center', 'bottom'),
    'Bond Street': (0, 15, 'center', 'bottom'),
    'Holborn': (0, 15, 'center', 'bottom'),
    'Chancery Lane': (0, 15, 'center', 'bottom'),
    'Hyde Park Corner': (0, 27, 'center', 'top'),

    
    'Goodge Street': (12, 0, 'left', 'center'),
    'Embankment': (12, -18, 'center', 'center'),
    'Victoria': (-12, 0, 'right', 'center'),
    'Pimlico': (-12, 0, 'right', 'center'),

    
    'Baker Street': (-12, 12, 'right', 'bottom'),
    "Regent's Park": (12, 12, 'left', 'bottom'),
    'Elephant & Castle': (12, -12, 'center', 'top'),
    'Covent Garden': (12, -12, 'left', 'top'),
    'Euston': (-12, 0, 'right', 'center')
}


for node, (x, y) in pos.items():
    dx, dy, ha, va = label_config[node]
    plt.annotate(
        node, xy=(x, y), xytext=(dx, dy), textcoords="offset points",
        fontsize=10.5, fontweight='bold', ha=ha, va=va,
        bbox=dict(boxstyle="round,pad=0.2", facecolor='white', edgecolor='none', alpha=0.9)
    )


edge_distances = nx.get_edge_attributes(G, 'dist')

edge_labels_with_units = {edge: f"{dist} km" for edge, dist in edge_distances.items()}

nx.draw_networkx_edge_labels(
    G, pos, edge_labels=edge_labels_with_units, 
    font_size=9, font_weight='bold', 
    bbox=dict(facecolor='white', edgecolor='none', alpha=1.0, pad=2)
)



legend_elements = [
    Line2D([0],[0], color='red', lw=4, label='Central'),
    Line2D([0],[0], color='blue', lw=4, label='Piccadilly'),
    Line2D([0],[0], color='black', lw=4, label='Northern'),
    Line2D([0],[0], color='brown', lw=4, label='Bakerloo'),
    Line2D([0],[0], color='lightblue', lw=4, label='Victoria')
]
plt.legend(handles=legend_elements, title="Lines Key", loc='upper right', frameon=True, fontsize=12, title_fontsize=13)

plt.margins(0.15)
plt.axis('off')
plt.show()
```

```python
import statistics

distances = [data['dist'] for _, _, data in G.edges(data=True)]

total_length = sum(distances)
average_distance = total_length / len(distances)

std_deviation = statistics.pstdev(distances) 

print("--- NETWORK STATISTICS RESULTS (TASK 3) ---")
print(f"1. Total length of the transport network: {total_length:.2f} kilometers")
print(f"2. Average distance between the stations: {average_distance:.2f} kilometers")
print(f"3. Standard deviation of the distances: {std_deviation:.4f} kilometers")
```
